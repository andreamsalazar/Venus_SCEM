```python
def solar_evolution(time):
    L_sun = 3.828e26 # W
    to = 4.57e9
    t_years = time*1
    Lum = L_sun*(1+(2/5)*(1-t_years/to))**(-1)
    S = Lum/(4*np.pi*a**2)
    return(S)
```

```python
def alpha_fit(S, Omega):
    sigma = Omega-n
    sigma_crit = 0.7e-6 # the thermal inertia timescale (for 1m ocean + atmosphere)
    #alpha_slow = 0.525+0.17*(np.log2(S/1366)) # have min and max scale with stellar radiation 
    alpha_slow = 0.03*(S-1257)**(1/3) + 0.28 + 0.08
    alpha_rapid = 0.34#+0.08*(np.log2(S/1366))
    alpha = alpha_rapid + (alpha_slow-alpha_rapid)/np.sqrt(1+(sigma/sigma_crit)**2) # based on dT 
    return(alpha)
```

```python
def Omega_crit_habitable(time, sign):
    S = solar_evolution(time)
    sigma_crit = 0.7e-6# transition between cloudy and not cloudy
    alpha_slow = 0.03*(S-1257)**(1/3) + 0.28 + 0.08
    alpha_rapid = 0.34#+0.08*(np.log2(S/1366))
    Omega_end_hab = n + sign*sigma_crit*np.sqrt(((alpha_slow-alpha_rapid)/np.maximum(1-alpha_rapid-1200/S,0))**2 -1)
    return(2*np.pi/(60*60*24*Omega_end_hab))
```

```python
from matplotlib.ticker import FuncFormatter
from matplotlib.ticker import MultipleLocator
def sample_trajectory(which):
    fig, ax = plt.subplots(2,2, dpi=150,figsize=(10,5), tight_layout=True)
    
    def rate_to_period(Omega, _):
        if Omega == 0:
            return("∞")
        period= 2*np.pi/Omega/(60*60*24)
        return(f"{period:.0f}")
    
    period_ticks = np.array([-1,-5,-50,-500,np.inf,500,50,5,1])
    omega_ticks = 2*np.pi/(period_ticks*60*60*24)
    #omega_ticks = omega_ticks[np.isfinite(omega_ticks)]
    for i in range(2):
        for j in range(2):
            rand_int = np.random.randint(0, len(which))
            color='k'
            ps = np.array(which['$p_s$'])[rand_int]
            ts = np.array(which['$t_s$'])[rand_int]/1e9
            where_hab = np.vstack(which['$h(t)$'])[rand_int]
            ax[i,j].plot(t_save/1e9, np.vstack(which['$\Omega(t)$'])[rand_int], '--', color=color, linewidth=2)
            ax[i,j].plot(t_save[where_hab>0.3]/1e9,np.vstack(which['$\Omega(t)$'])[rand_int][where_hab>0.3], '-', color=color, linewidth=3)
            ax[i,j].set_xlim(-0.2, 4.5)
            ax[i,j].set_ylim(-1e-4,1e-4)
            ax[i,j].set_yscale('symlog', linthresh=5e-7)
            ax[i,j].set_yticks(omega_ticks)
            ax[i,j].yaxis.set_major_formatter(FuncFormatter(rate_to_period))
            ax[i,j].minorticks_off()
            ax2 = ax[i,j].twinx()
            ax2.plot(t_save/1e9,np.vstack(which['$\epsilon(t)$'])[rand_int], '--', color='blue', linewidth=2)
            ax2.plot(t_save[where_hab>0.3]/1e9,(np.vstack(which['$\epsilon(t)$']))[rand_int][where_hab>0.3],'-', color='blue', linewidth=3)
            ax2.set_ylim(-10,190)
            #print(np.vstack(which['$\Omega(t)$'])[rand_int][where_hab>0.3][-1])
            ax2.set_yticks([0,30,60,90,120,150,180])
            ax2.set_ylabel('Obliquity (deg.)', color='blue')
            ax2.spines['right'].set_color('blue')
            ax2.tick_params(axis='y', colors='blue')
            #ax[i,j].annotate(str(np.round(ps,1)) + ' bar', (3, -5), fontsize=15)
            ax[i,j].grid(alpha=0.3)
            ax[i,j].set_ylabel('Rotation Period (days)')
            ax[i,j].set_xlabel('Time (Gyr)')
            ax[i,j].fill_between(t_eval/1e9, 2*np.pi/Omega_crit_habitable(t_eval,1)/(60*60*24), 100,color='red',alpha=0.1,linestyle='--',linewidth=0.5)
            ax[i,j].fill_between(t_eval/1e9, 2*np.pi/Omega_crit_habitable(t_eval,-1)/(60*60*24), -100,color='red',alpha=0.1,linestyle='--',linewidth=0.5)
            # major ticks every 10
            ax[i,j].xaxis.set_major_locator(MultipleLocator(1))

            # minor ticks at every point
            ax[i,j].xaxis.set_minor_locator(MultipleLocator(0.5))
            ax[i,j].axvline(ts, ymin=0,ymax=0.05,color='grey')
            ax[i,j].grid(True, which = 'both')
            #ax[i,j].set_xscale('log')
            #ax[i,j].axhline(y=2*np.pi/(60*60*24*225), color='orange', linewidth=1, zorder=0)
```

```python
def day_steam(varx, vary, minx, maxx, miny, maxy, labelx,labely, obl, freq):
    fig, ax = plt.subplots(2,dpi=300)
    plt.subplots_adjust(hspace=0.01)
    Venuscolor = 'darkorange'
    altVenuscolor = 'royalblue'
    TLcolor = 'silver'
    noteqcolor = 'black'
    ## tidally locked
    n=1
    size=[10, 4, 10,10]
    everhab = [TL_everhab, not_eq_everhab, altVenus_everhab, Venus_everhab]
    neverhab = [TL_neverhab, not_eq_neverhab, altVenus_neverhab, Venus_neverhab]
    prograde_zorder = [0, 3, 1, 2]
    retrograde_zorder = [0,3,2,1]
    everhab_alpha = 1
    neverhab_alpha = 1
    lw=0.3
    div=2
    everhab_line = 'w'
    neverhab_line='r'
    n_orb = 2*np.pi/(225*60*60*24)
    colors = [TLcolor, noteqcolor, altVenuscolor, Venuscolor]
    for i in range(len(everhab)):
        prograde_everhab = everhab[i].loc[everhab[i]['$\epsilon_o$'] < 90]
        retrograde_everhab = everhab[i].loc[everhab[i]['$\epsilon_o$'] >= 90]
        prograde_neverhab = neverhab[i].loc[neverhab[i]['$\epsilon_o$'] < 90]
        retrograde_neverhab = neverhab[i].loc[neverhab[i]['$\epsilon_o$'] >= 90]

        ax[0].scatter(prograde_everhab[varx][::n],prograde_everhab[vary][::n], facecolor = colors[i],edgecolors=everhab_line,lw=lw,alpha=everhab_alpha,s=size[i], zorder=prograde_zorder[i])
        ax[0].scatter(prograde_neverhab[varx][::n], prograde_neverhab[vary][::n], facecolor = colors[i],edgecolors=neverhab_line,lw=lw,alpha=neverhab_alpha,s=size[i], zorder=prograde_zorder[i])
        ax[1].scatter(retrograde_everhab[varx][::n], -retrograde_everhab[vary][::n], facecolor = colors[i],edgecolors=everhab_line,lw=lw,alpha=everhab_alpha,s=size[i], zorder=retrograde_zorder[i])
        ax[1].scatter(retrograde_neverhab[varx][::n], -retrograde_neverhab[vary][::n], facecolor = colors[i],edgecolors=neverhab_line,lw=lw,alpha=neverhab_alpha,s=size[i], zorder=retrograde_zorder[i])
        ax[0].set_ylim(400,0)
        ax[1].set_ylim(-0,-400)
        ax[0].get_xaxis().set_visible(False)
        ax[0].set_yticks([1,50,100,150,200,250,300,350])
        ax[1].set_yticks([-1,-50,-100,-150,-200,-250,-300,-350])
        
        ax[0].set_xscale('log')
        ax[0].set_xticks([])
        ax[1].set_xscale('log')
        ax[1].set_xlim(minx,maxx)
        ax[0].set_xlim(minx,maxx)
        #plt.ylabel(labely)
        ax[1].set_xlabel(labelx)
        ax[0].grid(False)
        ax[1].grid(False)
        
        fig.text(0.04, 0.5, labely, rotation = -270,va="center", fontsize=8)
```

```python
def scatters_pathway_hab(finalstate, colors,varx, vary, minx, maxx, miny, maxy, labelx,labely, obl, freq):
    plt.figure(dpi=200)
    Venuscolor = 'darkorange'
    altVenuscolor = 'royalblue'
    TLcolor = 'slateblue'
    noteqcolor = 'black'
    ## tidally locked
    n=1
    size=10#[15, 4, 15,15]
    pathways = np.unique(finalstate['pathway'])
    prograde_zorder = [0, 3, 1, 2]
    retrograde_zorder = [0,3,2,1]
    everhab_alpha = 1
    neverhab_alpha = 1
    lw=0.3
    div=2
    everhab_line = 'w'
    neverhab_line='r'
    n_orb = 2*np.pi/(225*60*60*24)
    #colors = 'black'#[TLcolor, noteqcolor, altVenuscolor, Venuscolor]
    alphas = np.linspace(0.2,1,len(pathways))
    if obl == True:
        for i in range(len(pathways)):
            path =finalstate.loc[finalstate['pathway'] == pathways[i]]
            plt.scatter(path[varx][::n], np.sign(np.cos(np.radians(path['$\epsilon_o$'])))[::n]*path[vary][::n], facecolor = colors,edgecolors=everhab_line,lw=lw,alpha=alphas[i],s=size, label = pathways[i])

    else:
        for i in range(len(everhab)):
            prograde_everhab = everhab[i].loc[everhab[i]['$\epsilon_o$'] < 90]
            retrograde_everhab = everhab[i].loc[everhab[i]['$\epsilon_o$'] >= 90]
            prograde_neverhab = neverhab[i].loc[neverhab[i]['$\epsilon_o$'] < 90]
            retrograde_neverhab = neverhab[i].loc[neverhab[i]['$\epsilon_o$'] >= 90]
            plt.scatter(prograde_everhab[varx][::n], prograde_everhab[vary][::n], facecolor = colors[i],edgecolors=everhab_line,lw=0.1,alpha=everhab_alpha,s=size, zorder=prograde_zorder[i])
            plt.scatter(prograde_neverhab[varx][::n], prograde_neverhab[vary][::n], facecolor = colors[i],edgecolors=neverhab_line,lw=0.1,alpha=neverhab_alpha,s=size, zorder=prograde_zorder[i])
            plt.scatter(retrograde_everhab[varx][::n], retrograde_everhab[vary][::n], facecolor = colors[i],edgecolors=everhab_line,lw=0.1,alpha=everhab_alpha,s=size, zorder=retrograde_zorder[i])
            plt.scatter(retrograde_neverhab[varx][::n], retrograde_neverhab[vary][::n], facecolor = colors[i],edgecolors=neverhab_line,lw=0.1,alpha=neverhab_alpha,s=size, zorder=retrograde_zorder[i])
    plt.ylim(miny,maxy)
    plt.ylabel(labely)
    plt.xlabel(labelx)
    plt.xlim(minx, maxx)
    plt.legend(fontsize=8)
    if varx == '$p_s$' or varx =='$t_s$':
        plt.xscale('log')
    if vary == '$p_s$' or vary =='$t_s$':
        plt.yscale('log')
        
    ## add arrows
    
```
